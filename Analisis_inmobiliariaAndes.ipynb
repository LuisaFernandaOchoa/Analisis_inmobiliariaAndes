# 📊 Proyecto: Análisis Comercial Inmobiliario

El sector inmobiliario necesita evaluar su **desempeño comercial** para comprender su crecimiento, rentabilidad y comportamiento de clientes.

Se solicita construir un **dashboard ejecutivo** que permita analizar ventas, clientes y propiedades para apoyar decisiones estratégicas basadas en datos.


# 📂 Descripción de los Datasets

🔹 **hecho_ventas_propiedades**  
Tabla de hechos con las transacciones de venta (precio, cliente, propiedad, canal, fecha).

🔹 **dim_clientes**  
Información y segmentación de clientes.

🔹 **dim_propiedades**  
Características de las propiedades (tipo, tamaño, ubicación, etc).

🔹 **dim_fecha**  
Tabla calendario que deberá ser creada como parte del modelado.


# 🎯 Preguntas que el dashboard debe responder

**Desempeño general**

- ¿Cuál es el **ingreso total** generado por la empresa?
- ¿Cuántas **propiedades se han vendido**?
- ¿Cuál es el **precio promedio de venta**?
- ¿Cuál es la **comisión total** generada por la empresa?

**Análisis comercial**

- ¿Qué **tipo de propiedad genera más ingresos**?
- ¿Qué **segmento de clientes compra más propiedades**?
- ¿Qué canal de venta **genera mayor ingresos**?


**Análisis temporal**

- ¿Cómo evolucionan las **ventas en el tiempo**?
- ¿El negocio está **creciendo año contra año**?
- ¿Cómo va el desempeño **acumulado del año actual (YTD)**?

**Cohortes de clientes**


- ¿Los clientes **vuelven a comprar** después de su primera compra?
- ¿Qué **cohortes de clientes generan más compras/ingresos con el tiempo**?

## 📊 Paso 4: Creación de medidas

En este paso deberás crear las **medidas necesarias para responder las preguntas de negocio del dashboard**.

Las medidas pueden implementarse en **Power BI (DAX)** o como **Calculated Fields en Tableau**.

---

### 4.1 🔹 Medidas base

Crear como mínimo las siguientes medidas:

- **Ingreso Total** (suma del `precio_venta`)
- **Cantidad de Ventas**
- **Ticket Promedio**
- **Comisión Total**

Estas métricas permitirán analizar el desempeño general del negocio.

---


### 4.2 🔹 Medidas con contexto de filtro

Debes crear **al menos 2 medidas** que modifiquen el contexto de filtro.

En este caso se utilizarán **medidas de participación (%)**, que permiten entender **qué proporción del ingreso total aporta cada categoría**.

Las medidas que deben construirse son:

- **Participación de ingresos por tipo de propiedad**
- **Participación de ingresos por canal de venta**
- **Participación de ingresos por segmento de cliente**

Para ello:
- Modifica el contexto de filtro con las funciones propias de cada herramienta
  - En Power BI, usa `CALCULATE`
  - En Tableau, usa expresiones como `FIXED` (LOD)

Elimina el filtro de la dimensión correspondiente para obtener el total

Asegúrate de que el denominador represente el total sin el filtro de la categoría, mientras que el numerador sí lo conserva.

Estas medidas se utilizarán principalmente como **tooltip o etiqueta adicional en los gráficos** del dashboard para complementar el análisis.


---

### 4.3 🔹 Inteligencia de tiempo

Utilizando la tabla **`dim_fecha`** (Power BI) o las funciones de fecha en Tableau, debes crear al menos 2 de las siguientes medidas:
- **Ventas Year to Date (YTD)**
- **Ventas Month to Date (MTD)**
- **Ventas del año anterior**
- **Crecimiento Year over Year (YoY)**

Estas métricas permitirán analizar la evolución del negocio en el tiempo.

---

### 4.4 🔹 Medidas para análisis de cohortes

Debes crear cálculos que permitan analizar **si los clientes vuelven a comprar después de su primera compra**.

⚠️ Estos cálculos deben crearse como **columnas calculadas**, no como medidas.  
Todas deben crearse en la tabla **`hecho_ventas_propiedades`**.

Debes crear como mínimo:

- **Primera compra por cliente** → fecha mínima de compra de cada cliente.
- **Mes Cohorte** → mes en que el cliente realizó su **primera compra** (mes de adquisición).
- **Mes Venta** → mes en que ocurre cada transacción. Este campo permite identificar **cuándo el cliente vuelve a comprar** después de su primera compra.

Estos cálculos permitirán construir una **matriz de cohortes** donde:
- **Filas** representen el **mes de cohorte**
- **Columnas** representen el **mes de venta**
- **Valores** muestren **cantidad de ventas o ingresos**

---

### Desarrollo paso 4.1 

Se realizó la construcción de las medidas a través de campo calculado. A continuación la construcción de cada métrica:
- Ingreso total = SUM([Precio Venta])
- Cantidad de ventas = COUNTD([Id Venta])
- Ticket promedio = [Ingreso total]/[Cantidad de ventas]
- Comision total = SUM([Monto Comision])

### Desarrollo paso 4.2 
- Participación de ingresos por tipo de propiedad = { FIXED [Tipo Propiedad (Dim Propiedades.Csv)] : [Ingreso total] } /{FIXED : [Ingreso total]}
- Participación de ingresos por canal de venta = {FIXED [Canal Venta]: [Ingreso total]}/{FIXED : [Ingreso total]}
- Participación de ingresos por segmento de cliente = {FIXED [Segmento Comprador]:[Ingreso total]}/{FIXED : [Ingreso total]}

### Desarrollo paso 4.3
Para dar respuesta a las preguntas de negocio:
- ¿Como evolucionan las ventas en el tiempo?: Se decidió que la pregunta puede ser respondida mediante un gráfico de líneas con fecha en Columnas e [Ingreso Total] en Filas, con esto se muestra la tendencia mes a mes directamente.
  
- ¿El negocio esta creciendo año contra año?: Para contrastar con el año anterior, se decide construir la medida `venta año anterior = LOOKUP([Ingreso Total], -12)`. Posteriormente se crea `Crecimiento YoY% = ([Ingreso total]-[Ventas año anterior])/[Ventas año anterior]`.

- ¿Como va el desempeño del año actual (YTD)?: Se construye `Acumulado YTD = RUNNING_SUM([Ingreso total])`, que posteriormente en la visualizacion se configuro para reiniciar por año utilizando la función Calcular usando.

### Desarrollo paso 4.4 
Se calcula la `Primera compra por cliente = { FIXED [Id Cliente]:MIN([Fecha Venta]) }`
Posteriormente, se calcula el mes de adquisición `Mes cohorte = DATETRUNC('month', [Primera compra por cliente])`
Para el mes de venta, se realiza siguiendo la estructura anterior con DATETRUNC y Fecha venta.
Se calculo también la edad de la cohorte como `Meses desde primera compra por cliente = DATEDIFF('month', [Primera compra por cliente], [Fecha Venta])`. Finalmente, se construye la Matriz de Cohortes (ingresos), según instrucciones del ejercicio.

## 📝Paso 6:  Resumen Ejecutivo

Escribe aquí un resumen con:

- hallazgos clave
- métricas principales
- insights accionables
- recomendaciones estratégicas

Ejemplos:

**Hallazgos clave**
- El ingreso total del periodo fue de $6,012,510,170, generados por 8,500 ventas.
- El tipo de propiedad que genera mayor Revenue es *Casa*, con un 37,26% de participación del ingreso, correspondiente a $2,240,535,304. Ninguna categoría supera el 40% del ingreso total, lo que evidencia una cartera de productos diversificada.
- La ciudad con mayor volumen e ingreso es Bogotá con 4,586 ventas equivalentes al 54% del ingreso total, frente a 3,914 ventas en Ciudad de México, correspondientes al 46% del ingreso total.
- El canal de ventas más eficiente en términos de ingresos es *Corredor*, que continúa liderando pese a mostrar una caída en su capacidad de captación de nuevos clientes. 

**Métricas principales**
- Ingreso Total: $6,012,502,170
- Cantidad de Ventas: 8,500 
- Ticket Promedio:  $707,353.20
- Comisión Total:  $200,627,166 

**Insights accionables**
- El segmento de clientes "Primera vez" concentra la mayor parte del ingreso, resultado coherente con la baja tasa de recompra identificada en el análisis de cohortes.
Las ventas muestran un crecimiento de aproximadamente 11% año contra año (comparando el acumulado 2024 vs. 2023), con variaciones mensuales de entre -1% y +27% — un crecimiento sostenido pero volátil mes a mes.
- El comportamiento de las cohortes sigue un patrón estacional marcado, con repuntes de actividad en marzo-abril y agosto-septiembre, y caídas notorias a mitad y fin de año, un patrón consistente con la estacionalidad típica del sector inmobiliario, más que con diferencias estructurales entre cohortes específicas.
- La captación de nuevos clientes alcanzó su punto máximo en marzo-abril de 2023 (~$295M) y no ha vuelto a alcanzar ese nivel, con una caída más pronunciada en el canal Corredor que en el canal Directo.
- La matriz de cohortes confirma que sí existe recompra, pero es minoritaria frente al ingreso de primera compra y se distribuye de forma dispersa en el tiempo, sin un ciclo corto y predecible.

**Recomendaciones estratégicas**
- Priorizar la comercialización de propiedades tipo Casa, sin descuidar Comercial y Departamento, dado que la cartera actual está bien diversificada y ninguna categoría domina de forma categorica.
- Fortalecer el canal de ventas Corredor, que presenta la mayor participación en el revenue, e investigar las causas de su reciente caída en captación para revertir la tendencia.
- Aprovechar la estacionalidad identificada (picos en marzo-abril y agosto-septiembre) para concentrar campañas de marketing y esfuerzos comerciales en los periodos de mayor actividad natural del mercado.
- Implementar estrategias de retención para mejorar la recompra, evitando que esta dependa del comportamiento espontáneo del cliente más que de iniciativas comerciales activas dirigidas a "Primera vez".
